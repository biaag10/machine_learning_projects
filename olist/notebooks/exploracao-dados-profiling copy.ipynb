```python
import pandas as pd
import pandera.pandas as pa
```

```python
from pathlib import Path

DATA_PATH = Path("../data/raw/olist_orders_dataset.csv")

df = pd.read_csv(DATA_PATH)
```

```python
COLUNAS_DATA = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for coluna in COLUNAS_DATA:
    df[coluna] = pd.to_datetime(df[coluna], errors="coerce")
```

```python
# COMPLETUDE

schema_completude = pa.DataFrameSchema({
    "order_id": pa.Column(
        str,
        nullable=False
    ),

    "customer_id": pa.Column(
        str,
        nullable=False
    ),

    "order_status": pa.Column(
        str,
        nullable=False
    ),

    "order_purchase_timestamp": pa.Column(
        pa.DateTime,
        nullable=False
    ),

    "order_estimated_delivery_date": pa.Column(
        pa.DateTime,
        nullable=False
    )
})

schema_completude.validate(df)

print("✓ COMPLETUDE: aprovada")
```

```python
# UNICIDADE
schema_unicidade = pa.DataFrameSchema({
    "order_id": pa.Column(
        str,
        nullable=False,
        unique=True
    )
})

schema_unicidade.validate(df)

print("✓ UNICIDADE: aprovada")
```

```python
# VALIDADE

STATUS_VALIDOS = [
    "delivered",
    "shipped",
    "canceled",
    "unavailable",
    "invoiced",
    "processing",
    "created",
    "approved"
]

schema_validade = pa.DataFrameSchema({
    "order_status": pa.Column(
        str,
        nullable=False,
        checks=pa.Check.isin(STATUS_VALIDOS)
    )
})

schema_validade.validate(df)

print("✓ VALIDADE: aprovada")

schema_ids = pa.DataFrameSchema({
    "order_id": pa.Column(
        str,
        nullable=False,
        checks=pa.Check.str_length(
            min_value=32,
            max_value=32
        )
    ),

    "customer_id": pa.Column(
        str,
        nullable=False,
        checks=pa.Check.str_length(
            min_value=32,
            max_value=32
        )
    )
})

schema_ids.validate(df)

print("✓ VALIDADE DOS IDs: aprovada")
```
